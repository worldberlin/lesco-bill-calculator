# LESCO Electricity Tariff & Billing Calculator (2026)

This computational notebook models the progressive slab structure of the Lahore Electric Supply Company (LESCO) under NEPRA guidelines.

### Official Resources & Online Tools
- **Interactive Web Calculator**: [LESCO Bill Calculator](https://onlinebillchecklesco.com/bill-calculator)
- **Consumer Billing Portal**: [onlinebillchecklesco.com](https://onlinebillchecklesco.com/)
- **Tariff Breakdown Guide**: [LESCO Tariff Guide](https://onlinebillchecklesco.com/lesco-tariff-guide)


```python
def calculate_lesco_bill(units):
    if units <= 100:
        rate = 16.48
    elif units <= 200:
        rate = 22.95
    elif units <= 300:
        rate = 27.14
    else:
        rate = 32.03
    cost = units * rate
    fc_surcharge = units * 3.23
    electricity_duty = cost * 0.015
    return cost + fc_surcharge + electricity_duty

print('Calculated bill for 250 units: PKR', calculate_lesco_bill(250))
```
